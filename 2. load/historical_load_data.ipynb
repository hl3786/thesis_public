```python
# uploads 
import pandas as pd 
import numpy as np 
from sklearn.linear_model import LinearRegression
import sqlite3
```

```python
conn = sqlite3.connect("pjm_load.db")
cur = conn.cursor()

cur.execute("""
CREATE TABLE IF NOT EXISTS pjm_rto_load (
    datetime_beginning_ept TEXT,
    mw REAL,
    year INTEGER
)
""")

conn.commit()
```

```python
conn = sqlite3.connect("pjm_hourly_load.db")
cur = conn.cursor()

# Drop old table if it exists
cur.execute("DROP TABLE IF EXISTS pjm_hourly_load")

# Recreate with correct column
cur.execute("""
CREATE TABLE IF NOT EXISTS pjm_hourly_load (
    datetime_beginning_ept TEXT,
    mw REAL,
    year INTEGER
)
""")
conn.commit()
conn.close()
```

```python
conn = sqlite3.connect("pjm_hourly_load.db")

for year in range(2012, 2026):
    filepath = f"Data/hrl_load_{year}.csv"
    df = pd.read_csv(filepath)

    # Parse timestamp
    df['datetime_beginning_ept'] = pd.to_datetime(df['datetime_beginning_ept'])
    # Keep only RTO rows
    df = df[df['zone'] == 'RTO']

    # Add the year (optional but useful)
    df['year'] = year

    # Insert ONLY the columns we want:
    df[['datetime_beginning_ept', 'mw', 'year']].to_sql(
        "pjm_hourly_load",
        conn,
        if_exists="append",
        index=False
    )

    print(f"Inserted year {year}, rows: {len(df)}")

conn.close()
```

```python
# Check output 
conn = sqlite3.connect("pjm_hourly_load.db")

# Read the first 10 rows
df_head = pd.read_sql_query("""
    SELECT *
    FROM pjm_hourly_load
    ORDER BY datetime_beginning_ept
""", conn)

print(df_head)

conn.close()
```

```python
conn = sqlite3.connect("pjm_load.db")
cursor = conn.cursor() 

create_max_daily_load_table = """
CREATE TABLE IF NOT EXISTS pjm_load_max_daily AS
SELECT *
FROM pjm_rto_load
WHERE (DATE(datetime_beginning_ept), mw) IN (
    SELECT DATE(datetime_beginning_ept), mw
    FROM pjm_rto_load); 
"""

cursor.execute(create_max_daily_load_table)
conn.commit()

# Preview new table
sample = cursor.execute("SELECT * FROM pjm_load_max_daily LIMIT 5;").fetchall()
for row in sample:
    print(row)

conn.close()
```
