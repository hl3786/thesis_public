```python
import sqlite3
import pandas as pd
```

```python
# Connect to your database
db_path = "load_cases.db"
conn = sqlite3.connect(db_path)

# 2️⃣ Preview the first few rows of one table (e.g., weather_0)
query = """
SELECT * FROM weather_0 
WHERE date >= '2023-01-01' AND date < '2024-01-01';
"""

df_test = pd.read_sql(query, conn)

print("\n2013 weather_0:")
print(df_test)

# Close connection when done
conn.close()
```

```python
simple_df = pd.DataFrame(df_test['load'])
simple_df.rename(columns={'load': 'Demand'}, inplace=True)
simple_df.index = simple_df.index + 1  # Start index at 1
simple_df.to_csv('demand_data.csv', index_label='Hour')
print(simple_df)
```
