```python
import pandas as pd
import glob
import os

# 1. Setup the path to your folder
# Find all Summer and Winter CSVs
csv_files = glob.glob("*.csv")

print(f"Found {len(csv_files)} files to process.")

for file_path in csv_files:
    # Read the CSV
    # We don't set an index_col here to ensure we keep the structure exactly as is
    df = pd.read_csv(file_path)
    
    # 2. Change $/MWh column for rows 2, 3, 4, and 5
    # iloc[1:5] selects index 1, 2, 3, and 4 (the 2nd, 3rd, 4th, and 5th rows)
    # We use .at or .loc to ensure we are modifying the dataframe correctly
    column_name = "$/MWh"
    
    if column_name in df.columns:
        df.iloc[0:4, df.columns.get_loc(column_name)] = 100000.0
        df.iloc[4:6, df.columns.get_loc(column_name)] = ''

        # 3. Save back to the same filename (overwrite)
        # index=False prevents pandas from adding an extra 'Unnamed: 0' column
        df.to_csv(file_path, index=False)
        print(f"Successfully updated: {os.path.basename(file_path)}")
    else:
        print(f"Warning: Column '{column_name}' not found in {file_path}")

print("\nAll files have been processed and overwritten.")
```
