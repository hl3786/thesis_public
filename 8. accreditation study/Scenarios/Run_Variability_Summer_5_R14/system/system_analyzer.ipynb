```python
import pandas as pd

# Your first two lines
gen_var = pd.read_csv('Generators_variability.csv')

# 1. Find all columns that contain the word 'conventional_hydroelectric' (ignoring case)
nuclear_cols = [col for col in gen_var.columns if 'conventional_hydroelectric' in col.lower()]

# 2. Calculate the mean (average) across all rows for just those columns
nuclear_averages = gen_var[nuclear_cols].mean()

# 3. Print the results
print(nuclear_averages)
```
