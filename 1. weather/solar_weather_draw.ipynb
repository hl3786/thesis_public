# Draw weaather for Solar

```python
import pandas as pd 
import sqlite3
```

```python
# Select PJM-wide weather data for the appropriate date range 
conn = sqlite3.connect("weather_data_2.db")

# Read the first 10 rows
df_head = pd.read_sql_query("""
    SELECT *
    FROM system_weather_hourly
    WHERE date >= "2007-01-01 00:00:00+00:00" AND date < "2013-12-30 01:00:00+00:00"

""", conn)

print(df_head)

conn.close()
conn.close()
```

```python
# Save output to CSV 
df_head.to_csv("solar_weather_data.csv")
```
