Add appropriate metadata to load_zone_results files

```python
import pandas as pd
import os

# 1. Define the exact metadata columns you want to bring over
metadata_cols = [
    'Voll', 'Demand_Segment', 'Cost_of_Demand_Curtailment_per_MW',
    'Max_Demand_Curtailment', '$/MWh', 'Rep_Periods',
    'Timesteps_per_Rep_Period', 'Sub_Weights', 'Time_Index'
]

# 2. Load the sample file and extract ONLY those metadata columns
sample_path = "inputs/Summer_1_sample.csv"
df_sample = pd.read_csv(sample_path)
df_metadata = df_sample[metadata_cols].copy()

# Ensure the metadata dataframe is exactly 168 rows long to match a 1-week file
# (This grabs the data + the NaNs that fill the rest of the column)
df_metadata = df_metadata.head(168)

# 3. Define the target folder
output_folder = "load_zone_results"
seasons = ["Summer", "Winter"]

# 4. Loop through all 20 of your generated files
for season in seasons:
    for i in range(1, 11):
        filename = f"{season}_{i}.csv"
        file_path = os.path.join(output_folder, filename)
        
        if os.path.exists(file_path):
            # Read the generated file (which currently just has your zones and thi)
            df_generated = pd.read_csv(file_path)
            
            # Combine the metadata and the generated data side-by-side
            # axis=1 concats them horizontally
            df_final = pd.concat([df_metadata, df_generated], axis=1)
            
            # Overwrite the file with the new complete layout
            df_final.to_csv(file_path, index=False)
            print(f"Successfully added metadata to {filename}")
        else:
            print(f"Warning: Could not find {file_path}")

print("\nAll files have been updated with metadata!")
```
